```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split as tts
```

```python
df = pd.DataFrame([[153,63], [174,81], [136,56],
                   [186,91], [128,47], [136,57],
                   [179,76], [163,72], [131,48]])

df.columns = ["x", "y"]
#202431089_Adhiwira Wildan Pratama
```

```python
df.corr()
```

```python
x_train = df['x'].values[:,np.newaxis]
y_train = df['y'].values

x_train
#202431089_Adhiwira Wildan Pratama
```

```python
y_train #202431089_Adhiwira Wildan Pratama
```

```python
lm = LinearRegression()
lm.fit(x_train, y_train)
#202431089_Adhiwira Wildan Pratama
```

## Prediksi

```python
x_test = [[170], [160], [180], [171], [150]]
p = lm.predict(x_test)
p
#202431089_Adhiwira Wildan Pratama
```

```python
# Koefisien dan intersep

print("Koefisien : ", lm.coef_)
print("Intersep : ", lm.intercept_)
#202431089_Adhiwira Wildan Pratama
```

```python
# visualisasi model

pb = lm.predict(x_train)
dfc = pd.DataFrame({"x" : df["x"], "y" : pb})
plt.scatter(df['x'], df['y'])
plt.plot(dfc['x'], dfc['y'], color="magenta", linewidth=1)
plt.xlabel('nilai X')
plt.ylabel('nilai Y')
#202431089_Adhiwira Wildan Pratama
```

```python
db_clean = df.dropna(subset=['umur', 'gaji', 'tinggi', 'berat', 'botak_prob'])

x = db_clean[['umur', 'gaji', 'tinggi', 'berat']]
y = db_clean['botak_prob']
#202431089_Adhiwira Wildan Pratama
```

```python
x_train, x_test, y_train, y_test = tts(x, y, test_size=0.2, random_state=0) #202431089_Adhiwira Wildan Pratama
```

```python
model = LinearRegression()

model.fit(x_train, y_train)
#202431089_Adhiwira Wildan Pratama
```

```python
y_pred = model.predict(x_test) #202431089_Adhiwira Wildan Pratama
```

```python
from sklearn.metrics import mean_squared_error, r2_score

print("Koefisien : ", lm.coef_)
print("Intersep : ", lm.intercept_)
print("MSE :", mean_squared_error(y_test, y_pred))
print("R2 Score : ", r2_score(y_test, y_pred))
#202431089_Adhiwira Wildan Pratama
```

```python
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha = 0.5)
plt.xlabel('nilai asli y_test')
plt.ylabel('nilai predilsi y_pred')
plt.title('nilai asli vs nilai prediksi')
plt.plot([0,1],[0,1],'r--')
plt.show()
#202431089_Adhiwira Wildan Pratama
```
